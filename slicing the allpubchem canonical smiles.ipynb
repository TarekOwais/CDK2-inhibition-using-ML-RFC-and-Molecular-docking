```python
import pandas as pd

# Load the entire dataframe
df = pd.read_csv('allpubchem_col15_canonicalsmiles58080608.csv')

```

```python
# Calculate number of rows per smaller dataframe
rows_per_df = 1000000

# Split the dataframe into smaller dataframes
split_dfs = []
for i, start in enumerate(range(0, len(df), rows_per_df)):
    # Extract a chunk of rows
    split_df = df.iloc[start:start+rows_per_df]
    
    # Append the chunk to the list of split dataframes
    split_dfs.append(split_df)
    
    # Write each split dataframe to a separate CSV file
    split_df.to_csv(f'allpubchem_part{i}.csv', index=False)
```
