```python
from chembl_webresource_client.new_client import new_client
import numpy as np, pandas as pd
```

```python
target=new_client.target.search("CDK2")
targets= pd.DataFrame.from_dict(target)
targets.head()
```

```python
targets.head(20)
```

```python
selected_target=targets.target_chembl_id[9]
selected_target
```

```python
activity=new_client.activity
res=activity.filter(target_chembl_id=selected_target)
df_all_types=pd.DataFrame.from_dict(res)
df_all_types.head()
```

```python
df_all_types.shape
```

```python
# saving the dataframe
df_all_types.to_csv("df_all_targets.csv", index=False, header= True)
```

```python
# selecting the IC50 only
res2=activity.filter(target_chembl_id=selected_target).filter(standard_type="IC50")
df=pd.DataFrame.from_dict(res2)
df.shape
```

```python
# saving
df.to_csv("df_ic50.csv", index=False, header= True)
```
